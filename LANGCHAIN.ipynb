```python
import os
from getpass import getpass
```

```python
from langchain_groq  import ChatGroq
from langchain_core.messages import *
```

```python
os.environ["GROQ_API_KEY"] = getpass("Enter GROQ API Key: ")
```

```python
llm = ChatGroq(model='openai/gpt-oss-120b')
response = llm.invoke("explain langchain in simple word to a beginner")
print(response.content)
```

```python
from langchain_core.prompts import *
```

```python
prompt = ChatPromptTemplate.from_template(
    """ Explain this {topic} in a very easy way in 5 sentences with real world example"""
    )
```

```python
user_input= input("what topic you want to learn")
print(user_input)
```

```python
formatted_prompt = prompt.invoke({"topic":user_input})
response = llm.invoke(formatted_prompt)
print(response.content)
```

```python
chains = prompt | llm
result = chains.invoke({"topic":user_input})
print(result.content)
```

```python
from langchain_community.document_loaders import CSVLoader
csvloader = CSVLoader("/content/sample_data/mnist_test.csv")
docs = csvloader.load()
print(len(docs))
print(docs[0].page_content)
```

```python
from langchain_text_splitters import *
splitter = RecursiveCharacterTextSplitter(chunk_size=100, chunk_overlap=0)
doc_splits = splitter.split_documents(docs)
print(len(doc_splits))
print(doc_splits[0].page_content)
```

```python
from langchain_core.tools import tool

def calculator(expression: str) -> str:
  """
  Find out the answer of the given expression.
  """
  return str(eval(expression))

calculator_tool = tool(calculator)

response2 = llm.invoke("2500*18/100")
print(response2.content)

calculator_tool.invoke({"expression":"2500*18/100"})
```

```python
llm_with_tools = llm.bind_tools([calculator_tool])
```

```python
response3 = llm_with_tools.invoke("calculate gst of 2500 at 18% rate")
print(response3.content)
```
